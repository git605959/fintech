# BeautifulSoup
* https://www.crummy.com/software/BeautifulSoup/bs4/doc/ 참고
* pip install beautifulsoup4
* from bs4 import BeautifulSoup as bs
* find, find_all 함수: xml, html에서 태그 기반으로 내용 찾음
* select, select_one 함수 : xml이나 html에서 css selector 기반으로 내용 찾음
* find_all, select는 해당 태그나 css select를 가진 부분을 모두 찾아서 list로 반환
* find_all, select는 여러 태그나 css select 중에서 가장 먼저 나오는 태그/selector를 한개만 찾아줌
* 찾아온 태그.name : 태그 이름 반환
* 찾아온 태그.text, .string: 태그 안쪽 텍스트 반환

```python
def extract_data(soup):
    s_data={}
    for item in soup.select('items'):
        for tags in item:
            try:
                for tag in tags:
                    if type(tag).__name__ == 'NavigableString':
                        pass
                    #print(type(tag))
                    else:
                        #print(tag.name,'&',tag.text)
                        s_data.setdefault(tag.name,[]).append(tag.text)
            except:
                pass
    s_df=pd.DataFrame(s_data)
    return s_df
```

# key_list를 만들고 select로 직접 추출하기

```python
page=1
df_list=[]
while True:
    url="https://apis.data.go.kr/1160100/service/GetSmallLoanFinanceInstituteInfoService/getOrdinaryFinanceInfo"
    payload=dict(serviceKey=serviceKey,numOfRows=1000,pageNo=page)
    r=requests.get(url,params=payload)
    print(r.url)
    print(r.status_code)
    r=r.content
    soup=bs(r,'xml')
    total_pages=int(soup.find('totalCount').text) // 100 +1
    key_list={}
    for item in soup.select("*"):
            key_list.setdefault(item.name,"")
    key_list = list(key_list.keys())
    result={}
    for key in key_list[10:]:
        for item in soup.select(key):
            result.setdefault(item.name,[]).append(item.text)
    df_list.append(pd.DataFrame(result))
    if page < total_pages :
        page +=1
    else:
        break
df_list
```

```python
df_final=pd.concat(df_list)
df_final=df_final.reset_index(drop=True)
df_final.to_csv("서민금융상품_대출정보.csv",encoding="utf-8",index=False)
```
