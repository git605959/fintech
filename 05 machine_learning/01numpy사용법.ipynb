# ndarray 만들기
* numpy의 기본 자료형 ndarray
* np.array(리스트/튜플)
* 자료형은 반드시 1가지만 넣는다.

* np.arange(시작,끝,step)

* np.sort

```python
arr=np.array([
    [[5,2,8],
    [1,6,7]],
    
    [[4,9,3],
    [0,2,1]]
])
print(arr)
print(arr.ndim)
print(arr.shape)
```

```python
# axis = -1 : 가장 안쪽 차원 1차원 배열을 정렬
print(arr)
np.sort(arr)
```

```python
# axis=1 : 열 별로 정렬 (1,2번째 3,4번째 줄 세로방향 비교)
np.sort(arr,axis=1)
```

```python
np.sort(arr,axis=0)
```

* np.where(조건,참,거짓) 3항 연산

# 배열 합치기 np.concatenate()

```python
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
print(a, a.shape)
print(b, b.shape)
```

* np.concatenate((a,b))
* 합치는 기본 방향은 axis=0 행이 늘어남 (아래로 붙음)
* axis=1 열이 늘어남(옆으로 붙음)

## 합치는 방향의 모양이 일치하지 않으면 합치기 X

```python
c=np.array([5,6])
print(c,c.ndim,c.shape)
```

```python
np.concatenate((a,c))
```

```python
d=np.array([[7,8,9]])
e=np.array([[1,2,3],[4,5,6]])
np.concatenate((d,e))
```

```python
f=np.array([[1,2],[3,4],[5,6]])
print(f)
```

```python
np.concatenate(e,f)
```

```python
# 전치행렬: 행과 열을 바꿈
# .T or .transpose()
print(f)
print(f.shape)
print(f.T)
```
