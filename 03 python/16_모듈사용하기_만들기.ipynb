# OS 모듈

```python
import os
```

```python
# 내가 사용하는 컴퓨터의 기본 정보 출력하기
print("현재 운영체제: ",os.name) #window=nt
print("현재 사용하는 폴더: ",os.getcwd()) #cwd=current working directory 
print("현재 폴더 내부의 파일 리스트: ",os.listdir())
```

```python
files=[]
for file_name in os.listdir():
    if file_name[-3:]=="txt":
        print(file_name)
        with open(file_name,encoding="utf-8") as file:
            data = file.read()
            files.append(data)
```

```python
files
```

```python
# 현재 위치에서 폴더 생성
os.mkdir("testdir")
```

```python
# 현재 위치에서 폴더 삭제
os.rmdir("testdir")
```

```python
# cmd 명령어 실행하기
os.system("dir")
```

```python
# 파일 생성 후 이름 변경
with open("original.txt","w") as file:
    file.write("안녕하세요")
os.rename("original.txt","new.txt")
```

```python
#폴더가 있는지 확인하고 없으면 폴더 생성하기
if not os.path.exists("./testdir"):
    os.mkdir("testdir")
else:
    print('폴더 존재함')
```
