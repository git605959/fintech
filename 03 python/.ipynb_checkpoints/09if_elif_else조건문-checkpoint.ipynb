elif:<br>

* 숫자를 입력받아 홀수인지 짝수인지 판단하는 프로그램

```python
x=input("숫자를 입력하세요")
if int(x) %2==0:
    print(f"입력한 숫자 {x}는 짝수입니다")
else:
    print(f"입력한 숫자 {x}는 홀수입니다")
```

```python
x=int(input())
if 0<=x<=10 and x %2==0:
    print(f"{x}는 0에서 10사이에 있는 숫자이고 짝수입니다")
else:
    print(f"{x}는 0에서 10사이에 있는 숫자가 아니거나 홀수입니다")
```

## 2. 다중 조건문 사용하기 elif
if 조건식:<br>
_ 조건식이 참일 때 실행할 코드<br>
elif 조건식1:<br>
_ 조건식1이 참일 때 실행할 코드<br>
elif 조건식2:<br>
_ 조건식2이 참일 때 실행할 코드<br>
else:<br>
_ 조건식이 모두 거짓일 때 실행할 코드<br>

```python
# 점수를 입력받아 81-100 A, 61-80 B, 41-60 C, 21-40 D, 0-20 F
score=int(input("점수를 입력하세요"))
if 0 <=score<=100:
    if 0<= score <=20:
        print(f"{score}는 F등급입니다.")
    elif 21<= score <=40:
        print(f"{score}는 D등급입니다.")
    elif 41<= score <=60:
        print(f"{score}는 C등급입니다.")
    elif 61<= score <=80:
        print(f"{score}는 B등급입니다.")
    elif 81<= score <=100:
        print(f"{score}는 A등급입니다.")
else:
    print("숫자를 0-100사이로 입력해주세요")
```

## 3. 조건 표현식

* print(결과값) if 참일 때 실행코드 else 거짓일 때 실행코드

```python
num = int(input("숫자를 입력하세요"))
print("짝수") if num%2==0 else print("홀수")
```
