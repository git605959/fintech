21.**courses라는 변수에 dict() 함수를 사용하여 다음과 같은 딕셔너리를 생성하세요.**
```python
{'programming': {'python': 50, 'java': 30, 'Cpp': 20},
 'design': {'Photoshop': 25, 'Illustrator': 15}}
```

````python
programming=dict({'python':50,'java':30,'Cpp':20})
design=dict({'Photoshop':25,'Illustrator':15})
courses=dict({'programming':programming, 'design':design})
courses
````

````python
courses = dict(programing=dict(python=50,java=30,Cpp=20),design=dict(Photoshop=25,Illustrator=15))
courses
````

````python
courses2=dict(programming=dict(zip(['python','java','Cpp'],[50,30,20])),design=dict(zip(['Photoshop','Illustrator'],[25,15])))
courses2
````
