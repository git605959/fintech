# 파이썬 기초 문법
* 파이썬의 주석은 문장 앞에 #을 붙여서 만듦
* 파이썬은 들여쓰기를 사용해서 코드를 구분해 작성 (코드블록)
* 들여쓰기는 4칸이 기본, google colab은 2칸
* 변수를 만들 때 스네이크 표기법 사용 ex) user_name (언더바로 연결)
* 클래스를 만들 때는 카멜 표기법 사용 ex) UserName, userName


## Print 문
* print 함수 안에 있는 내용을 출력
* print(출력할 내용)
* print(1) 출력할 내용이 숫자인 경우, 변수인 경우
* print('hello world') 출력할 내용이 문자, 문자열인 경우 반드시 따옴표가 필요
* 따옴표까지 같이 출력을 위해서는 섞어서
<br>    print("이것은 '파이썬'이다.")
<br>    print('이것은 "파이썬"이다.')

## print문의 옵션
* sep: 구분자 - print문 안에 여러개의 자료를 출력할 때 자료를 구분해주는 문자
* 기본값은 "" (공백)

* end: -print가 끝나고 실행되는 옵션
* 기본값 = \n (이스케이프 문자 new line)
    * \t (tap)
    * \r (carriage return 덮어쓰기)
    
## 함수의 옵션보기
* 함수의 옵션은 괄호 안에 커서를 놓고 shift+tap으로 볼 수 있음

## 변수
* 자료를 담는 택배상자
* 택배상자는 1개만 넣을 수 있음
* 변수는 영문과 숫자를 조합해서 만들 수 있음
* 변수명의 시작은 반드시 영문자로 시작 or _로 시작
* 변수명이 숫자로 시작하거나 -등의 특수문자로 시작하면 오류
* 예약어(이미 파이썬에서 사용하고 있는 함수명, 문법용 단어)는 사용하지 말 것
    * 주의!! 예약어를 써도 변수가 만들어지므로 예약어를 잘 기억해두어야 함
    * 추후 예약어 사용 시 에러 발생 가능
* 변수
* = 등호 오른쪽의 자료를 왼쪽 변수에 할당한다.

```python
print("이것은",'파이썬','입니다.',sep="_", end="\r")
print("이것은",'강아지','입니다.',sep="_")
```

```python
#shift+tap
print()
```
