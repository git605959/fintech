# 연산자의 종류
* 산술, 할당, 비교, 논리, 식별, 멤버

## 산술연산자 
* +,-,*,/,%,//,**

```python
print(1+1)
print(1-1)
print(1+3.1)
print(5/3)
print(4/2) #무조건 실수
print(3%2) # %2나눠서 홀짝 판별 시 사용
print(7//3)
print(2**3)
```

## 할당연산자
* =, +=, -=, *=, /=, %=

```python
a,b,c=10,10,10
d,e=7,4

a += 5
print(a)
b -= 5
print(b)
c *=10
print(c)
d/=2
print(d)
e%=3
print(e)
```

## 비교연산자
* ==: 같다
* !=: 다르다 #다르면 true, 같으면 false
* \> : 초과
* < : 미만
* \>= : 이상
* <= : 이하

```python
a = 3
b = 3.0
c = 5
d = "Python"
e = "python"
f = True
g = False
```

```python
a == b
```

```python
d==e #대소문자 구분
```

## 논리 연산자
* and, or, not
* and는 두 값이 모두 참이어야 참
* or는 두 값 중 하나라도 참이면 참, 두 값 모두가 거짓이어야 거짓
* not은 값을 반대로 뒤집음
* not > and > or 의 우선순위

```python
print(True and True)
print(True and False)
print(False and False)
```

```python
print(True or True)
print(True or False)
print(False or False)
```

```python
print(not False)
```

```python
not True or False and not False
```

## 식별연산자
* is, is not (object, data type)이 같은지 비교

```python
print(a)
print(b)
print(c)
```

```python
print(a == b) # a는 정수, b는 실수 3.0 이라서 값이 같음
print(a is b) # data type이 다르기 때문에 false로 판별
```

```python
print(a is not b) # True 면 다르다는 뜻
```

## 멤버연산자
* in, not in
* 리스트같은 자료형에 특정 요소가 포함되어 있는지를 판별

```python
mem = [1,2,3,4,5]
print(3 in mem)
print(10 in mem)
print(10 not in mem)
print(3 not in mem)
```
