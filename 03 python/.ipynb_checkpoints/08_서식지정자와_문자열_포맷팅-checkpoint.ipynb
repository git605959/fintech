# 서식지정자, format 메서드
* 문자열로만 입력해야하는 상황에서 같은 위치에 문자를 여러 번 바꿔야 할 때 사용
* 문자의 자릿수 맞춤, 위치 정렬에도 사용

## 1. 서식지정자 %d, %s, %f, %길이s

### 1) 문자를 입력할 때 %s

```python
print ("이 수업의 이름은 %s입니다" % " 국어")
```

```python
for i in ("국어","수학","사회"):
    print ("이 수업의 이름은 %s입니다" % i)
```

### 2) 정수를 입력할 때 %d

```python
print("올해는 %d년입니다." % 2025)
```

### 3) 실수를 입력할 때 %f 
* 기본 소숫점 6자리로 표시

```python
print("사람의 정상 체온은 약 %f 입니다." % 36.5)
print("사람의 정상 체온은 약 %.1f 입니다." % 36.5)
```

### 서식지정자로 문자열 길이, 정렬 %길이s

```python
print("%3s"%"01")
print(150000) #기본적으로 왼쪽 정렬
print("가격은 %10d입니다."%150000)
```

### 5) 여러 항목 동시에 넣기

```python
print("오늘은 %4d년 %d일" % (2025, 10)
```

```python
company=["삼성전자","LG전자","현대차","대한항공","하나투어"]
stock=[71800,121000,195000,31900,86600]
```

## 2. format 메서드 사용하기, f-string 문자열 포맷팅

### 1) format 메서드 사용하기
* "문자열 {인덱스} 문자열끝".format(값)

```python
print("{}의 주가는 {}입니다.".format('삼성전자',71800))
print("{0}의 주가는 {1}입니다.".format('삼성전자',71800))
print("{1}의 주가는 {0}입니다.".format('삼성전자',71800))
```

```python
for com,sto in zip(company, stock):
    print("{0}의 주가는 {1}원입니다.".format(com,sto))
```

### f-string으로 값넣기 (python3.6 사용 가능)

```python
company
```

```python
print(f"{company[0]}의 주가는 {stock[0]}입니다")
```

## 3.fomat 메서드와 f-string으로 문자열 정렬하기, 자리수 지정해 출력하기
* "(인덱스:<길이}".format(값)
* f"{값:<길이}"

```python
"{0:<10}".format("삼성전자")
```

```python
"{:>10}".format("삼성전자")
```

```python
"%10s" % "삼성전자"
```

```python
f"{'삼성전자':<10}"
```

* 숫자 표기 시 숫자 자리수 맞추고 기본 값 지정하기
* "{인덱스:채울 숫자 길이 서식지정자}".format(값), 단, 서식지정자에서 %는 빼고 적는다.
* f"{값:채울 숫자 길이 서식지정}"%는 뻄

```python
print("{0:02d}월".format(1))
print(f"{1:04d}월")
```

* 실수의 정수 자리수와 소수점 자리수 지정하기
* "{인덱스:기본값 자리수 서식지정자}".format(값)
* f"{값:기본값 자리수 서식지정자}"
* 실수에서 자리수는 정수 +,+소수점

```python
print("{0:03.3f}".format(2.245678))
print("{0:06.3f}".format(2.234567))
```

* 00시 00분부터 11시 59분까지 출력하세요. 1자리는 01형식으로 출력하세요.

```python
for i in range(12):
    print(i)
```
