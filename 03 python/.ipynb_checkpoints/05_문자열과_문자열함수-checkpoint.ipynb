## 인덱스와 문자열
* 문자 : 글자 1개
* 문자열: 글자가 여러개 => 순서가 있음 (앞(왼쪽)에서부터 0~)
* 순서가 있는 자료는 번호를 불러서 그 번호에 해당하는 자료를 꺼낼 수 있음
<br>
<br>
* 문자열의 길이를 잴 때 len() 함수 사용
* 문자열의 끝에서 n개를 가져오고 싶을 때 (오른쪽부터 문자열 슬라이싱)
    * 뒤에서부터 순서 : -1, -2, -3 (-1부터 시작)
* 슬라이싱 시 시작과 끝은 생략 가능
    * 모두 생략 시 처음부터 끝까지
* 슬라이싱 [시작위치:끝위치+1:건너뛸 문자 수(step)]
    * 역순정렬 : 문자열[::-1] #처음부터 끝까지, 거꾸로(-1) 정렬

```python
h4 = "문자열의 길이는 0부터 시작합니다."
h4[-11:-1]
```

```python
# h4에서 0번부터 10번까지 가져올 때
h4[-11:]
```

```python
h4='문자열은 왼쪽부터 번호를 매길 수 있으며 시작은 0번부터 입니다.'
print(h4[0::3]) #3개씩 건너뛰기
print(h4[::-1]) #역순정렬
```

## 문자열 함수

* .replace() : 값을 찾아서 바꾸는 함수
* .split() : 공백을 기준으로 자료를 나눠서 리스트로 변형
* .join() : 기준자를 통해서 리스트를 하나의 문자열로 변형
* .strip() : 문자열 외부의 공백제거, 공백+특수문자 제거

### replace 함수
* replace 함수는 사용 후 결과를 재할당 해야 함. (원본 자료는 바뀌지 않는 함수이기 때문)

```python
h5 = "Hello Python"
print(h5.replace('Python','world'))
print(h5) #원본은 변형 없음
```

### split 함수
* 구분자를 기준으로 자료를 나눠서 list로 만듦.
* 구분자 따로 지정 가능

```python
h6 = "문자열은, 왼쪽부터, 번호를, 매길, 수, 있으며, 시작은, 0번부터, 입니다."
splited_h6=h6.split(',')
splited_h6
```

### join 함수
* 리스트, 튜플을 하나의 문자열로 합치는 함수
* 합칠기준.join(리스트 or 튜플)
* 원본 변형 없음

```python
", ".join(h6) 
```

### 소문자 / 대문자 변환 함수
* upper(), lower()
* 원본 변형 없음

```python
h7 = "iron-man spider-man hulk"
h7=h7.upper()
print(h7)
h7 = h7.lower()
h7
```

### 왼쪽, 오른쪽, 양쪽 공백/특수문자 제거하기

* lstrip(): 왼쪽 공백 삭제
* rstrip(): 오른쪽 공백 삭제
* strip(): 양쪽 공백 삭제
* 제거하고 싶은 특정 문자를 () 안에 하나씩 넣어주면 됨


```python
h8 = "     공백이 있는 문자열     "
h8.lstrip()
h8.lstrip().replace(" ","") #공백 전부 삭제
h8.replace("     ","") #공백 n개씩 삭제 가능
```

```python
h9="...,  공백과 특수문자가 같이 있는 문자열 !! ^^ ㅋㅋㅋㅋ"
h9.strip("., !^ㅋ")
```

## method chaining (메서드 체이닝)
* 여러개의 메서드(함수)를 연결해서 사용하는 것


```python
h10 = "... ,  공백과 특수문자가 \n 같이 있는 \t 문자열 !! ^^ ㅋㅋㅋㅋ "
h10.strip("., !^ㅋ").replace('\n ',"").replace('\t ','')
```
