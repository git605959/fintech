# 리스트 표현식
* 리스트 안에 for 반복문과 if else 조건문을 한 번에 적어서 리스트를 만드는 방법
* 여러 줄의 코드를 1줄로 줄일 수 있음


```python
for i in range(1,11):
    print(f'{i}는 짝수입니다.') if i%2==0 else print(f'{i}는 홀수입니다.')
```

```python
[i for i in range(1,11)]
```

### 반복문과 표현식 실행 속도 차이

```python
import time

start=time.time()
a = []
for i in range(1,200000000):
    a.append(i+5)
print(a[:10])

print("코드 실행 시간: ", time.time() - start)
```

```python
start2=time.time()
b=[i+5 for i in range(1,200000000)]
print(b[:10])
print('코드 실행 시간:',time.time() - start2)
```

```python
import numpy as np
```

```python
start3=time.time()
c=np.arange(1,200000000)
c=c+5
print(c[:10])
print('코드 실행 시간:', time.time() - start3)
```

```python
# 1에서 10까지 숫자 중 짝수만 더해 리스트 만들기
l1 = [i for i in range(1,11) if i%2==0]
l1
```

# 파일 읽기 쓰기 open(), with open()
## 1) 파일에 문자열을 쓰거나 읽을 때 open 함수 사용
* 파일 저장
변수명 = open("파일이름", 모드(w:쓰기, r:읽기)<br>
변수명.write("저장할 내용")<br>
변수명.close()<br>
**close()를 안하면 계속 열려있음

```python
file=open("test.txt","w")
file.write("파일 만들기 연습")
file.close()
```

<br>변수1.close()

## with open
* with open은 close를 하지 않아도 자동으로 close 처리를 해줌


```python
with open("test2.txt","w") as file3:
    file3.write("with open을 사용하면 파일을 자동으로 닫아주어 편리하다.")
```

```python
with open("test2.txt","r") as file4:
    data = file4.read()
print(data)
```

# encoding에 따른 읽기 오류
* 운영체제나 (windows, linux, macOS), 한글/영어 윈도우에 따라서 인코딩되는 format이 달라짐.
* 인코딩에 따른 오류가 발생 가능
* web 표준은 utf-8이므로 인코딩은 utf-8로 하는 것이 좋다.
* windows의 경우 인코딩이 cp949이므로 주의

```python
with open("test_utf-8.txt","w",encoding='utf-8') as file5:
    file5.write('이것은 with open으로 만들 utf-8 인코딩 파일입니다.')
```

```python
with open("test_utf-8.txt","r") as file6:
    file6_data=file6.read() #window 기본 cp949로 읽어오기 때문에 에러 발생.
print(file6_data)
```

```python
with open("test_utf-8.txt","r",encoding="utf-8") as file6:
    file6_data=file6.read()
print(file6_data)
```

# json 파일 읽어와서 내용 정리하기


```python
with open("./data/서울특별시_관광지입장정보_2011_2016.json","r",encoding="utf-8") as file7: #상대경로 입력 후 tab키로 자동완성 가능
    trip_info=file7.read()
type(trip_info) #전체 문자열로 가져옴
```

## 문자열을 JSON/DICT 변환
* Json 모듈
* json.loads(함수)

```python
import json
```

```python
data = json.loads(trip_info)
data[0]
```

```python
for item in data[0].values():
    print(item)
```

```python
for item in data[:2]:
    for value in item.values():
        print(value)
# for key in item: print(item[key])
```

* 1. data 변수에 json.loads()로 문자열 읽어온 파일을 json 형식으로 변환
* 2. 구조 파악하기 (리스트, 딕셔너리)
* 3. data는 리스트 안에 67개의 딕셔너리 세트 저장
* 4. 각 딕셔너리에는 ['ForNum','NatNum','gungu','resNm','rnum','sido','yyyymm'] 형태로 저장되어있음
* 5. 67개 딕셔너리에서 하나의 key 당 67개의 value를 리스트 형태로 저장
* 6. pd 데이터프레임으로 정리

* value를 리스트로 만든 후 dict 함수로 딕셔너리로 만들어 데이터프레임으로 변환

```python
year_mon=[]
sido=[]
place=[]
foreigners=[]
total=[]

for item in data:
    foreigners.append(item['ForNum'])
    total.append(item['NatNum'])
    place.append(item['resNm'])
    sido.append(item['sido'])
    year_mon.append(item['yyyymm'])
    
tour_info=dict(년월=year_mon,시도=sido,장소=place,외국인=foreigners,전체방문객수=total)
print(tour_info)
```

```python
import pandas as pd
df = pd.DataFrame(tour_info)
df
```

* 딕셔너리를 바로 만들어서 데이터프레임으로 변환
* key, value를 직접 할당해서 딕셔너리 만들기

```python
result={}
for key in data[0]:
    value_list=[]
    for item in data:
        #print(key, item[key])
        value_list.append(item[key])
        result[key]=value_list
df2=pd.DataFrame(result)
df2
```

* setdefault 함수 이용해 만들기

```python
dict_result={}
for dicts in data:
    for key, value in dicts.items():
        #print(key, value)
        dict_result.setdefault(key,[]).append(value)
#print(dict_result)
df3=pd.DataFrame(dict_result)
df3
```
