클래스에 기능 추가히기

````python
class FourCal():
    def setdata(self, num1, num2):
        #self -> 인스턴스 (=사용자)
        self.num1=num1
        self.num2=num2
        print("self.num1",self.num1)
        print("self.num2",self.num2)
        
    #덧셈
    def add(self):
        result = self.num1 + self.num2
        return result
    #뺄셈
    def sub(self):
        result=self.num1-self.num2
        return result
    #곱셈
    def mul(self):
        result=self.num1*self.num2
        return result
    #나눗셈
    def div(self):
        result=self.num1/self.num2
        return result
        
````

````python
sam= FourCal()
sam.setdata(3,5)
````

````python
sam.add()
````

````python
sam.sub()
````

````python
sam.mul()
````

# 생성자
* 클래스를 실행해서 인스턴스를 만들 때 초기값을 받도록 해주는 메소드
* 생성자(constructor)는 객체(인스턴스)가 생성될 때 자동으로 호출되는 메소드
* 주로 초기값을 받을 때 사용.
* 클래스 내에서 메소드명으로 "\_\_init\_\_"이라고 씀


````python
class FourCal2():
    def __init__(self,num1,num2): #인스턴스 생성 시 같이 num1,num2 받도록
        self.num1=num1
        self.num2=num2
    
    def add(self):
        result=self.num1+self.num2
        return result
    def sub(self):
        result=self.num1-self.num2
        return result
    def mul(self):
        result = self.num1*self.num2
        return result
    def div(self):
        result=self.num1/self.num2
        return result
````

````python
gildong=FourCal2() #init 특성 때문에 초기값을 넣지 않으면 에러남
````

````python
gildong=FourCal2(3,5)
````

````python
gildong.add()
````

````python
sam = FourCal2(2,10)
sam.add()
````

````python
class FourCal3():
    def __init__(self, num1, num2):
        self.num1 = 0
        self.num2 = 0
    def add(self):
        result = self.num1 + self.num2
        return result
    def sub(self):
        result = self.num1 - self.num2
        return result
    def mul(self):
        result = self.num1 * self.num2
        return result
    def div(self):
        result = self.num1 / self.num2
        return result
````

````python
sally = FourCal3(4,6)
````

````python
sally.add()
````

````python
sally.div()
````

# 클래스의 상속
* 어떤 클래스를 만들 때 다른 클래스의 기능을 물려받는 것
* 기존 클래스에 기능을 추가할 때
* A클래스를 상속받아 B클래스를 생성한다고 할 때
* B클래스는 A클래스의 모든 기능을 사용할 수 있다.

````python
# FourCal2 을 상속받아 제곱 기능을 추가한 MoreFourCal 만들기
class MoreFourCal(FourCal2):
    def pow(self):
        result = self.num1 ** self.num2
        return result
````

````python
dul = MoreFourCal(2,3)
dul.pow()
````

````python
dul.sub()
````

# 메서드 오버라이딩(method overriding)
* 부모 클래스에서 상속한 메서드를 자식 클래스에서 동일한 이름으로 재정의

````python
kim = FourCal2(4, 0)
kim.add()
````

````python
kim.div()
````

* FourCal2는 num2에 0이 입력될 경우 div메서드 실행시 0으로 나누게 되기 때문에 에러발생
* SafeFourCal를 만들고 FourCal2상속받은 후 div 함수를 재정의 하여 에러를 없앰

````python
class SafeFourCal(FourCal2):
    def div(self):
        if self.num2 == 0:
            return 0
        else:
            return self.num1 / self.num2
````

````python
kim = SafeFourCal(4, 0)
kim.div()
````

# 메서드 오버로딩
* 한 클래스 안에서 동일한 메서드의 기능을 다르게 재정의
* 동일한 이름의 메서드가 기능이 약간 변ㅇ경된 채로 여러 개 존재
* def add(num1,num2)
* def add(num1,num2,num3)
* def add(num1,num2,num3,num4)

````python
class OverloadingEx():
    def add(self,x,y):
        return x+y
    def add(self,x,y,z):
        return x+y+z
    def add(self,x,y,z,a):
        return x+y+z+a
````

````python
sam=OverloadingEx() #일반적으로는 맨 마지막 함수만 인식
sam.add(10,20,30,40)
````

# 정적메서드 static method
* 인스턴스를 만들지 않고도 클래스의 메서드를 바로 사용할 수 있는 메서드
* 메서드 위에 @staticmethod라는 데코레이터를 붙여서 만든다.
* static method는 self를 받지 않으므로 인스턴스 속성, 인스턴스 메서드가 필요하지 않을 때 사용

````python
class Cal():
    @staticmethod
    def add(a,b):
        print(a+b)
    @staticmethod
    def sub(a,b):
        print(a-b)
````

````python
sam=Cal()
sam.add(2,3)
````

````python
sally=Cal()
sally.add(10,3)
````

# 클래스 메서드 class method
* 메서드 위에 @classmethod를 붙여서 만든다.
* 클래스 메서드의 첫번째 매개변수에는 cls를 지정
* 클래스 메서드도 staticmethod처럼 인스턴스 생성 없이 호출 가능
* 클래스 메서드는 메서드 안에서 클래스속성, 클래스 메서드에 접근해야 할 때 사용


````python
class Factory():
    n_coffee_machine=0 #클래스 속성
    
    def __init__(self):
        Factory.n_coffee_machine +=1 #인스턴스가 생성될 때 클래스 속성에 1을 더함
    @classmethod
    def print_n_coffee_machine(cls):
        print(f'{cls.n_coffee_machine}개가 제조되었습니다.')
````

````python
Factory.print_n_coffee_machine()
````

````python
sam=Factory()
sam.print_n_coffee_machine()
````

# 추상 클래스 abc(abstract base class)
* abc 모듈 필요
* 추상클래스는 추상클래스를 상속받아 클래스를 만들 때 반드시 만들어야 하는 메소드를 지정해주는 클래스
* 인터페이스
* 규칙을 부여해서 그 규칙에 따라 클래스를 만들도록 강제 함
* 추상 클래스를 상속받은 클래스는 반드시 추상클래스에서 정의한 메소드를 overriding으로 구현해야 함.
* 이력서 양식에 맞추어 내용적기와 같다.

````python
from abc import *
````

````python
# 추상 클래스 - 필요한 클래스 틀만 짜서 다른 사람에게 줌
class CoffeeBase(metaclass=ABCMeta):
    @abstractmethod       #형식만 주고 내용은 pass
    def bean_input(self):
        pass
    
    @abstractmethod
    def grind(self):
        pass
    
````

````python
sam=CoffeeBase()
````

````python
class CoffeeMachine(CoffeeBase):
    def bean_input(self):
        print('커피 원두 넣기')
````

````python
sam = CoffeeMachine()
````

````python
# 추상 클래스를 상속받아서 필요한 클래스의 기능 추가
class CoffeeMachine2(CoffeeBase):
    def bean_input(self):
        print('커피 원두 넣기')
    def grind(self):
        print("커피 원두 갈기")
````

````python
sam = CoffeeMachine2()
````

````python
sam.bean_input()
````

````python
sam.grind()
````
