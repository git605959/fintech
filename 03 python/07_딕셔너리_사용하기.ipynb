### update(key=value)
* 기존에 key가 있는 경우 값을 새로운 값으로 update
* 기존에 key가 없는 경우 새로운 key, value 생성
* key 부분은 "," 사용하지 않는다.

```python
d.update(방문자수=500)
d.update(바람='강함')
d
```

### 딕셔너리에서 value를 가져오는 함수 .get()
* .get(key, key가 없을 때 기본 값)
* 딕셔너리 내에 key가 있다가 없다가를 반복하는 경우 오류 방지

```python
#d['사람'] #딕셔너리에 없는 key를 호출하면 키에러 발생
d.get('사람') # 없는 키 호출 시 error 방지
```

### 딕셔너리에서 key, value, (key, value)를 가져오는 함수
* keys(): 딕셔너리에서 key만 출력
    * list()함수로 형변환 해주면 list로 사용 가능
* values() : 딕셔너리에서 value만 출력
    * list()함수로 형변환 해주면 list로 사용 가능 (반복문 사용 가능)
* items() : 딕셔너리의 key와 value를 튜플 묶음 형태로 출력(key, value)
    * list() 형변환을 통해 list 처럼 사용 가능

```python
print(d.keys())
print(d.values())
print(d.items())
```

```python
list(d.items())[0]
```
